```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
import pandas as pd
import time
import random

from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC


target_url = "https://www.jumia.com.ng/catalog/productratingsreviews/sku/XI363MP7VZ6XSNAFAMZ/"
PRODUCT_NAME = "Redmi_A5"

MAX_REVIEWS = 1000
REVIEWS_PER_PAGE = 10
MAX_PAGES = MAX_REVIEWS // REVIEWS_PER_PAGE
CSV_FILE = "jumia_reviews_dataset.csv"


def scrape_jumia_reviews(url):
    options = webdriver.ChromeOptions()

    options.add_argument(
        "user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
        "AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120 Safari/537.36"
    )

    driver = webdriver.Chrome(
        service=Service(ChromeDriverManager().install()),
        options=options
    )

    print(f"Robot connecting to: {url}")

    all_reviews = []

    for page in range(1, MAX_PAGES + 1):
        page_url = f"{url}?page={page}"
        print(f"Scraping page {page}")

        driver.get(page_url)
        time.sleep(random.uniform(2.5, 4.5))

        try:
            WebDriverWait(driver, 10).until(
                EC.presence_of_all_elements_located((By.TAG_NAME, "article"))
            )
        except:
            print("No more reviews found.")
            break

        articles = driver.find_elements(By.TAG_NAME, "article")

        if not articles:
            break

        for article in articles:
            if len(all_reviews) >= MAX_REVIEWS:
                break

            try:
                title = article.find_element(By.TAG_NAME, "h3").text
                body = article.find_element(By.TAG_NAME, "p").text
                date = (
                    article.find_element(By.CLASS_NAME, "-d-co").text
                    if article.find_elements(By.CLASS_NAME, "-d-co")
                    else "N/A"
                )

                all_reviews.append({
                    "Review_Title": title,
                    "Review_Content": body,
                    "Date": date,
                    "Product_Name": PRODUCT_NAME
                })
            except:
                continue

        if len(all_reviews) >= MAX_REVIEWS:
            break

    driver.quit()

    if all_reviews:
        df = pd.DataFrame(all_reviews)

        with open(CSV_FILE, "a", encoding="utf-8", newline="") as f:
            df.to_csv(f, header=f.tell() == 0, index=False)

        print(f"SUCCESS: Appended {len(df)} reviews to {CSV_FILE}")
    else:
        print("FAILED: No reviews scraped.")


scrape_jumia_reviews(target_url)
```
